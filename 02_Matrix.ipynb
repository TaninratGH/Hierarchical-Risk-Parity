If you look at the columns when importing asset prices, you get a vector in $\mathbb{R}^n$. When you calculate returns from time $t_k \to t_{k+1}$, you lose one row of data(from the beginning). This means each asset now has a vector in $\mathbb{R}^{n-1}$.  the data changes its matrix dimension from the price matrix $\mathbf{M}_{m \times n}$ to the return matrix $\mathbf{M}_{(m-1) \times n}$ (for example, from $\mathbf{M}_{43 \times 7}$ to $\mathbf{M}_{42 \times 7}$).

```python
import pandas as pd

def return_matrix(prices: pd.DataFrame) -> pd.DataFrame:
    
    returns = prices.pct_change().dropna()

    return returns

prices = pd.read_csv("prices.csv")
prices = prices.drop(columns=["Date"], errors="ignore")
return_matrix(prices)

```

```python
Cov = prices.cov()
Cov
```

```python
import numpy as np
Cov = Cov.to_numpy()
eigenvalues, P = np.linalg.eig(Cov)
D = np.diag(eigenvalues).real
```

```python
print("Diagonal matrix (D):")
display(pd.DataFrame(D))
```

```python
Corr = prices.corr()
Corr = Corr.to_numpy()
eigenvalues_p, P = np.linalg.eig(Corr)
D = np.diag(eigenvalues_p).real
print("Diagonal matrix (D):")
display(pd.DataFrame(D))
```

```python
dist = np.sqrt(2 * (1 - Corr))
display(pd.DataFrame(dist))
eigenvalues_p, dist = np.linalg.eig(dist)
```
