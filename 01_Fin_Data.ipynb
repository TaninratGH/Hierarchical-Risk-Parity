```python
import pandas as pd
import yfinance as yf

tickers = [
    "AAPL", "MSFT", "NVDA", "AMZN",
    "GOOGL", "META", "TSLA"
]

# 2-month data
data = yf.download(
    tickers,
    start="2026-08-09",
    end="2026-10-10",
    interval="1d",
    auto_adjust=True,
    progress=False
)

prices = data["Close"]
prices.index.name = "Date"

prices.to_csv("prices.csv")

# Now I get only Close price into the folder - prices.csv
```

```python
import pandas as pd
import yfinance as yf

tickers = [
    "AAPL", "MSFT", "NVDA", "AMZN",
    "GOOGL", "META", "TSLA"
]

# 2-month data
data = yf.download(
    tickers,
    start="2026-08-09",
    end="2026-10-10",
    interval="1H",
    auto_adjust=True,
    progress=False
)

prices = data["Close"]
prices.index.name = "Date"

prices.to_csv("hourly_prices.csv")
```
